```python
import numpy as np
import pandas as pd
```

```python
df1=pd.read_csv("chicago.csv")
df2=pd.read_csv("New_Orleans.csv")
```

```python
df1["City"]="Chicago"
df2["City"]="New_Orleans"
```

```python
df=pd.concat([df1,df2],ignore_index=True)
```

```python
df.head(2)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop(['neighbourhood_group', 'license'], axis=1, inplace=True)
```

```python
df.dropna(subset=['price'], inplace=True)
```

```python
df['host_name'].fillna(df['host_name'].mode()[0], inplace=True)
```

```python
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')
df['last_review']=df['last_review'].fillna(df['last_review'].median())
```

```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(df['reviews_per_month'].mean())
```

```python
df.isnull().sum()
```

```python
df.to_csv("chicago_and_Orlean.csv")
```
